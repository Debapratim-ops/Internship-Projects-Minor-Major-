Importing dependencies

```python
import numpy as np
import pandas as pd
```

Loading the data set into pandas 

```python
Airquality_data = pd.read_csv('C:\DEKSTOP FILES\Python internship programs\Major Project\Data Set\AirQualityUCI.csv', sep = ';', decimal = ',')
```

Printing First Five

```python
Airquality_data.head()
```

Droping the last two columns using splicing

```python
Airquality_data = Airquality_data.iloc[:, :-2]
Airquality_data.head()
```

```python
Airquality_data.tail()
```

Removing the Null Values from Data set( Last row containing data = 9356 )

```python
Airquality_data = Airquality_data.head(9357)
Airquality_data.tail()
```

```python
Airquality_data.info()
```

According to the UCI Documantation the actual null values are denoted as -200. 
Finding the number of null values and handling them 

```python
Airquality_data.isin([-200]).sum(axis = 0)
```

```python
Airquality_data = Airquality_data.replace(to_replace = -200, value = np.nan)
Airquality_data.tail()
```

```python
Airquality_data.isnull().sum()
```

```python
Airquality_data.mean(numeric_only=True)
```

```python
Airquality_data = Airquality_data.fillna(Airquality_data.mean(numeric_only=True))
Airquality_data.tail()
```

```python
Airquality_data.isnull().sum()
```

```python
# Converting the date from DD/MM/YYYY to YYYY-MM-DD format
Airquality_data['Date'] = pd.to_datetime(Airquality_data['Date'], format='%d/%m/%Y')
Airquality_data.head()
```

```python
time_info = Airquality_data['Time']
date_info = Airquality_data['Date']
```

```python
print(type(time_info))
print(type(date_info))
```

Combining the two Series

```python
time_info = time_info.apply(lambda x:x.replace('.',':'))
date_time = pd.concat([date_info, time_info], axis = 1)
date_time.head()
```

```python
date_time['ds'] = date_time['Date'].astype(str) + ' ' + date_time['Time'].astype(str)
date_time.head()
```

```python
data = pd.DataFrame()
data['ds'] = date_time['ds']
data['y'] = Airquality_data['RH']
data.head()
```

Forecasting the Air Quality using FB Prophet Algorithm for (Relative Humidity), FB Prophet Documentation - https://facebook.github.io/prophet/docs/quick_start.html

```python
from prophet import Prophet
```

```python
model = Prophet()
model.fit(data)
```

```python
future = model.make_future_dataframe(periods = 365, freq = 'D')
future.tail()
```

```python
forecast = model.predict(future)
forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail()
```

```python
fig1 = model.plot(forecast)
fig2 = model.plot_components(forecast)
```
